```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

DATA_PATH=r"/content/adult.csv"
df=pd.read_csv(DATA_PATH)
```

```python
df.drop_duplicates(inplace=True)
```

```python
df.replace("?", np.nan, inplace=True)
```

```python
cols=df[['age','educational-num','capital-gain','capital-loss','hours-per-week']]
```

```python
cols
```

```python
for col in cols:
  sns.boxplot(x=df[col])
  plt.show()
```

```python
for x in cols:
  Q1=df[x].quantile(0.25)
  Q3=df[x].quantile(0.75)
  IQR=Q3-Q1
  lowerboundary=Q1-1.5*IQR
  upperboundary=Q3+1.5*IQR
  df[x]=df[x].clip(lowerboundary,upperboundary)

```

```python
for col in cols:
  sns.boxplot(x=df[col])
  plt.show()
```

```python
cat_cols = [
    'workclass',
    'education',
    'marital-status',
    'occupation',
    'relationship',
    'race',
    'gender',
    'native-country'
]
```

```python
df=pd.get_dummies(data=df,columns=cat_cols,dtype=int,drop_first=True)
```

```python
df
```

```python
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()
df['income'] = le.fit_transform(df['income'])
```

```python
df
```

```python
x=df.drop('income',axis=1)
```

```python
x
```

```python
y=df['income']
```

```python
y
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
x_train.shape
```

```python
x_test.shape
```

```python
y_test.shape
```

```python
y_train.shape
```

```python
from sklearn.preprocessing import StandardScaler
sc=StandardScaler()
x_train=sc.fit_transform(x_train)
x_test=sc.transform(x_test)
```

```python
from sklearn.linear_model import LogisticRegression
lr=LogisticRegression(max_iter=1000)
lr.fit(x_train,y_train)
```

```python
y_pred=lr.predict(x_test)
```

```python
y_pred
```

```python
from sklearn.metrics import accuracy_score

baseline_test = accuracy_score(y_test, y_pred)

print("Test Accuracy:", baseline_test)
```

```python
from sklearn.model_selection import cross_val_score

cv_scores = cross_val_score(
    lr,
    x_train,
    y_train,
    cv=5,
    scoring='accuracy'
)

baseline_cv = cv_scores.mean()

print("CV Scores:", cv_scores)
print("Mean CV Accuracy:", baseline_cv)
```

```python
from sklearn.model_selection import GridSearchCV

params = {
    'C': [0.01, 0.1, 1, 10]
}

grid = GridSearchCV(
    LogisticRegression(max_iter=1000),
    param_grid=params,
    cv=5,
    scoring='accuracy'
)

grid.fit(x_train, y_train)
```

```python
print("Best Parameters:", grid.best_params_)
print("Best CV Accuracy:", grid.best_score_)
```

```python
grid_pred = grid.predict(x_test)

grid_test = accuracy_score(y_test, grid_pred)

print("Grid Test Accuracy:", grid_test)
```

```python
from sklearn.model_selection import RandomizedSearchCV

random_params = {
    'C': [0.001, 0.01, 0.1, 1, 10, 100]
}

random_search = RandomizedSearchCV(
    LogisticRegression(max_iter=1000),
    param_distributions=random_params,
    n_iter=4,
    cv=5,
    scoring='accuracy',
    random_state=42
)

random_search.fit(x_train, y_train)
```

```python
print("Best Parameters:", random_search.best_params_)
print("Best CV Accuracy:", random_search.best_score_)
```

```python
random_pred = random_search.predict(x_test)

random_test = accuracy_score(y_test, random_pred)

print("Randomized Test Accuracy:", random_test)
```

```python
from sklearn.ensemble import RandomForestClassifier

rf = RandomForestClassifier(
    n_estimators=100,
    random_state=42
)
```

```python
rf_scores = cross_val_score(
    rf,
    x_train,
    y_train,
    cv=5,
    scoring='accuracy'
)

rf_cv = rf_scores.mean()

print("Random Forest CV Accuracy:", rf_cv)
```

```python
rf.fit(x_train, y_train)
```

```python
rf_pred = rf.predict(x_test)
```

```python
rf_test = accuracy_score(y_test, rf_pred)

print("Random Forest Test Accuracy:", rf_test)
```

```python
results = pd.DataFrame({
    'Model': [
        'Baseline Logistic Regression',
        'GridSearchCV',
        'RandomizedSearchCV',
        'Random Forest'
    ],

    'CV Accuracy': [
        baseline_cv,
        grid.best_score_,
        random_search.best_score_,
        rf_cv
    ],

    'Test Accuracy': [
        baseline_test,
        grid_test,
        random_test,
        rf_test
    ],

    'Best Params': [
        'Default',
        str(grid.best_params_),
        str(random_search.best_params_),
        'Default'
    ]
})

results
```
